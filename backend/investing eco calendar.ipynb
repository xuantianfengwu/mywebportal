```python
def access_jin10_calendar(date_str):
    """
    访问金十数据指定日期的日历页面
    :param date_str: 日期字符串，格式如 "2025-11-24"
    """
    # 1. 配置Chrome浏览器选项
    chrome_options = Options()
    # 可选：添加无头模式（不显示浏览器窗口）
    # chrome_options.add_argument("--headless=new")
    # 可选：禁用图片加载，提升速度
    chrome_options.add_experimental_option("prefs", {"profile.managed_default_content_settings.images": 2})
    # 禁用浏览器提示
    chrome_options.add_argument("--disable-infobars")
    chrome_options.add_argument("--disable-blink-features=AutomationControlled")
    # 设置用户代理（避免被识别为爬虫）
    chrome_options.add_argument(
        "user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
    )

    # 2. 初始化WebDriver（自动下载匹配的Chrome驱动）
    driver = webdriver.Chrome(
        service=Service(ChromeDriverManager().install()),
        options=chrome_options
    )

    try:
        # 3. 访问目标网址
        url = f"https://rili.jin10.com/day/{date_str}"
        driver.get(url)
        
        # 4. 等待页面加载（可选：根据需要调整等待时间）
        driver.implicitly_wait(10)  # 隐式等待10秒
        time.sleep(2)  # 强制等待2秒确保页面完全加载

        # 5. 验证页面是否成功加载（示例：打印页面标题）
        print(f"页面标题: {driver.title}")
        print(f"当前URL: {driver.current_url}")

        # 可选：获取页面源代码
        page_source = driver.page_source
        # 可在这里添加数据提取逻辑（如查找特定元素、文本等）
    except Exception as e:
        print(f"访问出错: {e}")
    finally:
        # 6. 关闭浏览器（如果是无头模式，必须关闭否则进程残留）
        input("按回车键关闭浏览器...")  # 暂停查看结果
        driver.quit()

# 访问2025-11-24的日历页面
RESP = access_jin10_calendar("2025-12-22")
```

```python
from selenium.webdriver.common.by import By  # 必须导入
from selenium.webdriver.support.ui import WebDriverWait
import time

weekday_btns = driver.find_elements(By.CSS_SELECTOR, "div.jin-layout-content__left > div > div.index-page-header__wrap > div.date-slider__wrapper > div.date-slider > ul > li")
page_sources = [] # 5天的page_source
countdown_text = '' # 多久公布下一跳数据，用于侧面监测切换到新的日期页面了
for i,btn in enumerate(weekday_btns[:5]):
    print(i+1)
    btn.click()
    while True:
        time.sleep(3)
        try:
            new_countdown_text = driver.find_element(By.CSS_SELECTOR, "#jinTable1 > div.jin-table-body__wrapper > div > div > div > div.countdown-line").text
        except:
            new_countdown_text='no data'
        if countdown_text != new_countdown_text:
            page_sources.append(driver.page_source)
            countdown_text = new_countdown_text
            break
```

```python
from bs4 import BeautifulSoup

result1 = []
result2 = []
    
for page_source in page_sources:
    soup = BeautifulSoup(page_source)
    date = soup.select_one('div.datepicker-wrapper > div.datepicker-trigger > span').get_text(strip=True)
    
    for row in soup.select("#jinTable1 > div.jin-table-body__wrapper > div > div > div > div.jin-table-row"):
        content = row.get_text(strip=True, separator='|')
        result1.append( f"{date}|{content}")
    
    for row in soup.select("#jinTable2 > div.jin-table-body__wrapper > div > div.jin-table-row__group"):
        content = row.get_text(strip=True, separator='|')
        result2.append( f"{date}|{content}")
```

```python
# Data [['date', 'time', 'country', 'star', 'time_period', 'name']]
import pandas as pd
new_result1 = []
for r in result1:
    items = r.split('|')
    if len(items)==9:
        if items[-5]!='视频解析':
            items = items[:len(items)-4] + ['视频解析'] + items[-4:]
    if len(items)==10:
        ts = items[1]
    elif len(items)<10:
        items = items[:1] + [ts] + items[1:]
    new_result1.append('|'.join(items))

res_df1 = pd.DataFrame([r.split('|')[:3] for r in new_result1], columns=['date', 'time', 'name'])

fin_data_df = pd.read_excel('C:\codes\mywebapp\sqlite\jin10_financial_data_list.xlsx')
q_data_names = fin_data_df[fin_data_df['is_store']=='Y']['数据类型'].tolist()
def check_name(x):
    for n in q_data_names:
        if n in x:
            return True
    return False
data_df = res_df1[res_df1['name'].apply(lambda x: check_name(x))]
data_df
```

```python
# Data [['date', 'time', 'country', 'star', 'time_period', 'name']]
data_df['date'] = data_df['date'].apply(lambda x:x.replace('-',''))
def get_country_name(x):
    country_names = ['美国','日本','加拿大','英国','瑞士','中国', '法国', '德国','欧元区','澳大利亚']
    for n in country_names:
        if n in x:
            return n
    return ''
data_df['country'] = data_df['name'].apply(lambda x:get_country_name(x))
def get_time_period(x):
    time_period_names = [ '11月', '12月', '1月','2月','3月','4月','5月', '6月', '7月', '8月', '9月', '10月',
                         '第一季度','第二季度','第三季度','第四季度']
    for n in time_period_names:
        if n in x:
            return n
    return ''
data_df['time_period'] = data_df['name'].apply(lambda x:get_time_period(x))
data_df['star'] = 3

data_df [['date', 'time', 'country', 'star', 'time_period', 'name']]
```

```python
# Event [['date', 'time', 'country', 'star', 'region', 'people', 'event_content']]
event_df = pd.DataFrame([r.split('|')[:4] for r in result2], columns=['date', 'time', 'country', 'event_content'])
event_df['date'] = event_df['date'].apply(lambda x:x.replace('-',''))
event_df['region'] = ''
event_df['people'] = ''
event_df['star'] = 3

event_df[['date', 'time', 'country', 'star', 'region', 'people', 'event_content']]
```

```python
import pandas as pd
from openpyxl import load_workbook

def overwrite_excel_sheet(df, excel_path, sheet_name):
    """
    覆盖写入 Excel 文件的特定工作表

    Parameters:
    df: 要写入的 DataFrame
    excel_path: Excel 文件路径
    sheet_name: 要覆盖的工作表名称
    """
    # 读取现有 Excel 文件的所有工作表
    with pd.ExcelFile(excel_path) as xls:
        sheet_names = xls.sheet_names

    # 创建 ExcelWriter，模式为 'a'（追加/修改）
    with pd.ExcelWriter(excel_path, engine='openpyxl', mode='a', if_sheet_exists='replace') as writer:
        # 如果工作表已存在，会被覆盖
        # 如果工作表不存在，会被创建
        df.to_excel(writer, sheet_name=sheet_name, index=False)

    print(f"已成功将 DataFrame 写入到 {sheet_name} 工作表")
```
